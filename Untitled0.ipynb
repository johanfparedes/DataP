```python
# ==============================================================================
# 1. IMPORTAR LIBRERÍAS Y SUBIR ARCHIVO EXCEL
# ==============================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from google.colab import files # Librería específica de Colab para subir archivos

# Configuración de estilo para los gráficos
sns.set_theme(style="whitegrid")

print("Por favor, sube tu archivo Excel (ej. data.xlsx):")
uploaded = files.upload()
```

```python
# ==============================================================================
# 2. CARGAR Y EXPLORAR LOS DATOS
# ==============================================================================
# Cambia 'data.xlsx' por el nombre exacto de tu archivo si es diferente
nombre_archivo = list(uploaded.keys())[0] # Toma el nombre del archivo recién subido

# Leemos el archivo Excel
df = pd.read_excel(nombre_archivo)

# Si el Excel tiene una columna 'ID', la pasamos al índice
if 'ID' in df.columns:
    df.set_index('ID', inplace=True)

print(f"Dataset cargado correctamente. Dimensiones: {df.shape[0]} filas y {df.shape[1]} columnas.")
display(df.head()) # 'display' crea una tabla interactiva y bonita en Colab
```

```python
# ==============================================================================
# 3. PREPARACIÓN Y ESTANDARIZACIÓN DE DATOS
# ==============================================================================
print("--- ESTANDARIZACIÓN DE VARIABLES ---")
scaler = StandardScaler()

# Transformamos y mantenemos el formato de DataFrame
df_scaled = pd.DataFrame(scaler.fit_transform(df), columns=df.columns, index=df.index)

print("Datos estandarizados correctamente (Media=0, Varianza=1). Primeras filas:")
display(df_scaled.head())
```

```python
# ==============================================================================
# 4. MATRIZ DE CORRELACIÓN
# ==============================================================================
plt.figure(figsize=(10, 6))
sns.heatmap(df_scaled.corr(), annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.title("Matriz de Correlación - Evidencia de Multicolinealidad", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# 5. CÁLCULO DE COMPONENTES PRINCIPALES (PCA)
# ==============================================================================
pca = PCA()
pca.fit(df_scaled)

# Varianza explicada
varianza_explicada = pca.explained_variance_ratio_ * 100
varianza_acumulada = np.cumsum(varianza_explicada)

df_varianza = pd.DataFrame({
    'Componente': [f'CP {i+1}' for i in range(len(varianza_explicada))],
    'Autovalor (Eigenvalue)': pca.explained_variance_,
    'Varianza Explicada (%)': varianza_explicada,
    'Varianza Acumulada (%)': varianza_acumulada
})

print("--- RESUMEN DE VARIANZA EXPLICADA ---")
display(df_varianza.round(2).head(4)) # Mostramos los 4 primeros

# Gráfico de Sedimentación
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(varianza_explicada) + 1), varianza_acumulada, marker='o', linestyle='--', color='b', label='Varianza Acumulada')
plt.bar(range(1, len(varianza_explicada) + 1), varianza_explicada, alpha=0.6, color='g', label='Varianza Individual')
plt.axhline(y=80, color='r', linestyle='-', alpha=0.3, label='Línea del 80%')
plt.title("Scree Plot: Retención de Información por Componente", fontsize=14)
plt.xlabel("Número de Componente")
plt.ylabel("Porcentaje de Varianza Explicada")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# 6. INTERPRETACIÓN EJECUTIVA (CARGAS FACTORIALES - LOADINGS)
# ==============================================================================
print("\n--- CARGAS FACTORIALES (LOADINGS) DE LOS 2 PRIMEROS COMPONENTES ---")
# Extraemos cómo contribuye cada variable original a los 2 primeros componentes
loadings = pd.DataFrame(
    pca.components_.T[:, :2],
    columns=['CP1 (Perfil Valor vs Digital)', 'CP2 (Residual)'],
    index=df.columns
)
display(loadings.style.background_gradient(cmap='coolwarm', axis=0))
# Usamos un degradado de color para que sea más visual en Colab
```

```python
# ==============================================================================
# 7. SEGMENTACIÓN DE CLIENTES EN EL NUEVO ESPACIO
# ==============================================================================
# Transformamos los datos originales a las puntuaciones del PCA
pca_scores = pca.transform(df_scaled)
df['Score_CP1'] = pca_scores[:, 0]
df['Score_CP2'] = pca_scores[:, 1]

plt.figure(figsize=(12, 8))
sns.scatterplot(x='Score_CP1', y='Score_CP2', data=df, s=100, color='royalblue', alpha=0.7)

# Añadir etiquetas a algunos clientes (usando una muestra para no saturar el gráfico)
for i in range(0, len(df), 5):
    plt.text(df['Score_CP1'].iloc[i]+0.1, df['Score_CP2'].iloc[i]+0.1, df.index[i], fontsize=9)

# Líneas divisorias
plt.axvline(x=0, color='red', linestyle='--', alpha=0.5)
plt.axhline(y=0, color='red', linestyle='--', alpha=0.5)

# Interpretación sugerida en el gráfico
plt.text(2, 1.5, "Perfil 'Premium'\n(Altos Ingresos/Tickets)", fontsize=11, color='darkgreen', weight='bold')
plt.text(-4, 1.5, "Perfil 'Cazaofertas'\n(Alta Frecuencia/Digital)", fontsize=11, color='darkorange', weight='bold')

plt.title("Mapa de Segmentación de Clientes basado en CP1 y CP2", fontsize=15)
plt.xlabel(f"Componente Principal 1 ({varianza_explicada[0]:.1f}% de varianza)")
plt.ylabel(f"Componente Principal 2 ({varianza_explicada[1]:.1f}% de varianza)")
plt.tight_layout()
plt.show()

print("\n================ CONCLUSIÓN AUTOMATIZADA ================")
print(f"El PCA permitió reducir el conjunto de variables a 2 componentes principales ")
print(f"que explican de manera conjunta el {varianza_acumulada[1]:.1f}% del comportamiento del cliente.")
```

Interpretación Ejecutiva
Respondiendo a las preguntas estratégicas de la sesión:

1. ¿Cuántas variables se redujeron?
Se pasó de 8 variables operativas a solo 2 componentes estratégicos. Este proceso permitió eliminar el ruido y la redundancia en los datos, logrando conservar más del 95% de la información vital sobre el comportamiento del cliente.

2. ¿Qué representan los nuevos componentes?
Componente 1 (CP1) — Perfil de Compra: Premium Tradicional vs. Digital Cazaofertas

Puntaje Positivo (Perfil Premium): Clientes de mayor edad, altos ingresos y alta antigüedad. Tienen un ticket de compra muy elevado (monto promedio), pero compran con poca frecuencia. No suelen utilizar canales digitales ni reaccionan a promociones (Ej. Clientes ID 2, 21).

Puntaje Negativo (Perfil Digital/Volumen): Clientes más jóvenes y de menores ingresos. Compran muy seguido y prueban muchos productos distintos, pero con un ticket bajo. Son altamente dependientes de la página web y de las ofertas (Ej. Clientes ID 7, 29).

Componente 2 (CP2) — Factor de Exploración y Lealtad Residual

Aunque explica una menor proporción de la varianza (13.5%), este componente agrupa ligeras variaciones entre el tiempo que llevan interactuando con la marca (antigüedad) y la cantidad de productos diferentes que se atreven a probar, actuando de forma independiente a su edad o nivel de ingresos.

3. ¿Qué decisiones estratégicas puede tomar la empresa?
Para el segmento Positivo (Premium): Se recomienda detener el envío masivo de correos con descuentos, ya que no reaccionan a este estímulo ni al entorno web. La estrategia debe girar hacia la atención personalizada, catálogos físicos, o la creación de un programa de lealtad "VIP" basado en el valor del ticket, no en la frecuencia de visita.

Para el segmento Negativo (Digital/Volumen): Se debe optimizar la experiencia de usuario (UX) en la plataforma web. Es el público ideal para lanzar campañas flash, cupones digitales y estrategias de venta cruzada (cross-selling), dado que son muy sensibles al precio e interactúan constantemente con el ecosistema digital de la marca.

4. ¿Qué clientes se deben priorizar?
La priorización dependerá directamente del objetivo comercial (OKR/KPI) del trimestre en curso:

🎯 Si el objetivo es Rentabilidad y Flujo de Caja: Priorizar a los clientes con alto puntaje positivo en CP1 (perfil Premium de alto ticket).

🎯 Si el objetivo es Tráfico Web, Rotación de Inventario y Adquisición: Priorizar a los clientes con puntaje negativo en CP1 (perfil de alta frecuencia y volumen).
